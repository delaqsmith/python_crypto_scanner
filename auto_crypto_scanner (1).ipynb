```python
# Import BeautifulSoup for parsing and extracting data from HTML webpages
from bs4 import BeautifulSoup

# Import requests for sending HTTP requests to websites
import requests

# Import pandas for creating and working with DataFrames and CSV files
import pandas as pd

# Import datetime for recording the date and time when the data is collected
from datetime import datetime

# Import os for checking whether files or directories exist
import os

# Import time for adding delays or scheduling repeated execution of the script
import time
```

```python

def automated_bitcoin_tracker():
    url = 'https://coinmarketcap.com/currencies/bitcoin/'  # Stores the URL of the Bitcoin page
    
    page = requests.get(url)  # Sends a request and stores the response
    
    soup = BeautifulSoup(page.text, 'html') # Parse the HTML content of the webpage using BeautifulSoup

    # Find the Bitcoin name.
    # recursive=False ensures that only the direct text of the span is returned,
    # preventing the nested "price" text from being included.
    crypto_name = soup.find('span', class_='sc-c1554bc0-0 hhUhBo').find(string=True, recursive=False) 

    # Find and extract the current Bitcoin price from the webpage
    crypto_price = soup.find('span', class_= 'sc-c1554bc0-0 RbQXx base-text').text

    # Remove the dollar sign from the price
    final_price = crypto_price.replace('$','')

    # Get the current date and time when the data is collected
    date_time = datetime.now()

    # Create a dictionary containing the crypto name,
    # current price, and timestamp
    dict = {'Crypto Name':crypto_name,
            'Price':final_price,
           'TimeStamp': date_time}

    # Convert the dictionary into a Pandas DataFrame
    # The brackets create a single-row DataFrame
    df = pd.DataFrame([dict])

     # Check whether the CSV file already exists
    if os.path.exists(r'C:\Users\soosh\OneDrive\Desktop\Bitcoin Web Puller\Crypto_Automated_Pull.csv'):

        # If the file exists, append the new Bitcoin data
        # header=False prevents the column headers from being repeated
        # index=False prevents Pandas from adding row numbers
        df.to_csv(r'C:\Users\soosh\OneDrive\Desktop\Bitcoin Web Puller\Crypto_Automated_Pull.csv', mode='a', header = False, index = False)

    # If the file does not exist, create it and include column headers
    else:
        df.to_csv(r'C:\Users\soosh\OneDrive\Desktop\Bitcoin Web Puller\Crypto_Automated_Pull.csv', index = False)

# Display the collected Bitcoin data
    print(df)
```

```python
# Continuously run the Bitcoin tracker
while True:
    
    # Call the function to retrieve and save the current Bitcoin price
    automated_bitcoin_tracker()

    # Wait 10 seconds before running the function again
    time.sleep(10)
```
